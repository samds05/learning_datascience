```python
#Indentation, not braces.

#Unlike other languages, python follows white paces 
#to build structure code
sum = 0
for x in range(8):
    for y in range(x):
        sum+=y
print(sum)
#Here it follows whitespaces to make block of code
#executable it specified manner.
#A block of code follows same whitespaces till
#the end of block
```
